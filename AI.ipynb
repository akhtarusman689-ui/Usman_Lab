```python
destination = input("Enter your destination Dubai/Turkey/Malaysia: ")
budget = int(input("Enter your budget: "))

if destination.lower() == "dubai" and budget >= 100000:
    print("You can travel to Dubai.")
elif destination.lower() == "turkey" and budget >= 80000:
    print("You can travel to Turkey.")
elif destination.lower() == "malaysia" and budget >= 60000:
    print("You can travel to Malaysia.")
else:
    print("Your budget is not enough for this destination.")
```
